```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Load dataset
data = pd.read_csv("Heart_Disease_Prediction.csv")
data
```

```python
# Remove extra spaces
data.columns = data.columns.str.strip()
```

```python
X = data[['Age', 'BP', 'Cholesterol']].values
```

```python
# Convert Heart Disease → Target
# Absence = 0
# Presence = 1

def convert_disease(disease):
    if disease == 'Absence':
        return 0       # No Heart Disease
    else:
        return 1       # Heart Disease

y = data['Heart Disease'].apply(convert_disease).values
```

```python
# Create ANN model
model = Sequential()

# Input layer + Hidden layer
model.add(Dense(8, input_dim=3, activation='relu'))

# Output layer
model.add(Dense(2, activation='softmax'))
```

```python
# Compile model
model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
# Train model
model.fit(X, y, epochs=50)
```

```python
# Take user input
age = float(input("Enter Age: "))
bp = float(input("Enter Blood Pressure: "))
cholesterol = float(input("Enter Cholesterol: "))
```

```python
# Convert into numpy array
test = np.array([[age, bp, cholesterol]])

# Predict
pred = model.predict(test)

# Get predicted class
labels = ["Absence", "Presence"]

print("Heart Disease:", labels[np.argmax(pred)])
```
